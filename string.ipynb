```python
# 1. basic string operations
# 1.1 string creation
# Single or double quotes — same thing
name = 'Alice'
lang = "Python"

# Triple quotes for multi-line strings
msg = """Hello,
World!"""

# Check the type and length
print('name:', name)
print('type of name: ', type(name))   # <class 'str'>
print('length of name: ',len(name))    # 5

# Concatenate strings with +
greeting = "Hello, " + name
print('greeting', greeting)     # Hello, Alice

# Repeat with *
print("ha" * 3)      # hahaha
```

```python
# 1.2 Indexing & slicing
s = "Python"
#    P  y  t  h  o  n
#    0  1  2  3  4  5  (positive)
#   -6 -5 -4 -3 -2 -1  (negative)

print('first char of s: ',s[0])       # P  
print('last char of s: ',s[-1])      # n   

# Slicing [start:end] — end is excluded
print('first 3 chars of s: ',s[0:3])     # Pyt
print('3rd char to the end of s: ',s[2:])      # thon  (to end)
print('chars until the fourth char of s: ',s[:4])      # Pyth  (from start)

# Step — every Nth character
print(s[::2])     # Pto
print(s[::-1])    # nohtyP  (reversed!)
```

```python
# 1.3 commom methods
s = "  Iowa State University, located in Ames, is a renowned public research university.  "
# Case transformations
uppercase = s.upper()
lowercase = s.lower()
title = s.title()
print("UPPERCASE:", uppercase)
print("lowercase:", lowercase)
print("title:", title)
print()

# Strip whitespace
print("remove whitespace:",s.strip())      
print("remove right whitespace:",s.rstrip())    
print("remove left whitespace:",s.lstrip())   
print()

#split by space & join words using space
words = s.split()
print("Words in the sentence:", words)
clause = s.split(', ')
print("Clauses in the sentence:", clause)
print()

joined_sentence = " ".join(words)
print("Joined Sentence by space:", joined_sentence)
joined_sentence = "-".join(words)
print("Joined Sentence by -:", joined_sentence)
print()

#find substrings & replace substrings
index = s.find("Ames") 
# returns the index of the first occurrence of the substring. 
# if the substring is not found, it returns -1
print(f"'Ames' found at index: {index}")
print()

modified_sentence = s.replace("Ames", "Ames, Iowa")
print("Modified Sentence:", modified_sentence)
print()
```

```python
# 1.4 formatting
name = "Alice"
age = 30
pi = 3.14159

# f-strings (recommended)
print(f"Hello, {name}! You are {age}.")
# Hello, Alice! You are 30.

# Expressions inside f-strings
print(f"{name.upper()} is {age * 2} in dog years")
# ALICE is 60 in dog years

# Format numbers
print(f"Pi ≈ {pi:.2f}")   # Pi ≈ 3.14
print(f"{1000000:,}")      # 1,000,000
```

```python
# 2. Regular Expression (Regex)
import re

# 4.1 Check if a Pattern Exists
pattern = r"ames"
match = re.search(pattern, s)  # stops after finding the first match in the string
if match:
    print(f"Pattern '{pattern}' found at position: {match.start()}")
else:
    print(f"Pattern '{pattern}' not found")

pattern2 = r"\w+"
match = re.search(pattern2, s)  # stops after finding the first match in the string
if match:
    print(f"Pattern '{pattern2}' found at position: {match.start()}")
else:
    print(f"Pattern '{pattern2}' not found")
```

```python
# 4.2 Find all  Matching
pattern = r"university"
matches = re.findall(pattern, s, re.IGNORECASE) #Case-Insensitive
print(f"Case-insensitive matches for '{pattern}':", matches)

pattern2=r'\d+'
matches = re.findall(pattern2, 'room 42, floor 7, unit 100')
print(f"number matches for '{pattern}':", matches)         #find all numbers in string
```

```python
# 4.3 Split String Using a Pattern
pattern = r",|\."  # The pipe symbol "|" means "or" in regex; "\." matches the dot "."
parts = re.split(pattern, s)
print("Split Sentence:", parts)
print()
```

```python
# 4.4 Validate Patterns (e.g., Email-Like Text)
test_string = "Contact me at:qli@iastate.edu"

pattern = r"\b[A-Za-z0-9._-]+@[A-Za-z0-9_-]+\.[A-Za-z]{2,}\b"
# \b
# Matches a word boundary, ensuring the email address is a standalone word and not part of a larger string.
# This is used at both the start and end of the pattern.

# [A-Za-z0-9._-]+
# Matches the local part of the email address (before the @).
# Allows any combination of uppercase and lowercase letters (A-Za-z), digits (0-9), dots (.), underscores (_), and dashes (-).
# The "+" ensures there is at least one character.

# [A-Za-z]{2,}
# Matches the top-level domain (e.g., com, org, net).
# Accepts at least two characters ({2,}) and ensures they are only uppercase (A-Z) or lowercase (a-z) letters.

if re.search(pattern, test_string):
    print("Valid email found!")
    print(f'{re.search(pattern, test_string)}')
else:
    print("No valid email found.")
print()

# This pattern will fail for qli@company.co.uk
# how to change the pattern?
```

```python
# 4.5 replace matches
result = re.sub(r'\d+', '#', 'room 42, floor 7')
print(result)          # replace number with "#"

# sub with a function
result = re.sub(r'\d+', lambda m: str(int(m.group()) * 2), 'x=3, y=7')
print(result)          # double the numbers
```

```python
# 4.6 group 
# change name format
pattern = re.compile(r""" # re.compile will compile the pattern for repeated use.
    ( [A-Za-z'-]+ )       # Group 1 — last name (letters, hyphens, apostrophes)
    ,\s+                  # literal comma + one or more spaces
    ( [A-Za-z'-]+ )       # Group 2 — first name
    \s+                   # one or more spaces
    ( [A-Za-z] )          # Group 3 — middle name initial (first letter only)
    ( [A-Za-z'-]* )       # Group 4 — rest of middle name (captured, then discarded)
""", re.VERBOSE)          # re.VERBOSE (also written as re.X) just changes how Python reads the pattern string. more human readable patterns

replacement = r"\2 \3. \1"       # firstName middleInitial. lastName

names = [
    "Smith, John William",
    "O'Brien, Mary Catherine",
    "Garcia-Lopez, Carlos Miguel",
    "Van Der Berg, Emma Louise",
]
re_name=[]
for name in names:
    re_name.append(pattern.sub(replacement, name))

print(re_name)
```

```python
# Named groups — (?P<name>pattern)
# change date format

pattern=re.compile('(?P<year>\d{2,4})-(?P<month>\d{1,2})-(?P<day>\d{1,2})')

dates = [ #yyyy-mm-dd
    'Date: 2024-3-15',
    'Date: 2026-09-15',
    'Date: 89-2-3', #wrong date but right format
    ]
new_dates =[]
for date in dates:
    m = re.search(pattern, date)
    print(m.groupdict())    
    new_dates.append(pattern.sub(r'\g<month>/\g<day>/\g<year>',date))

print(new_dates)
 
```
